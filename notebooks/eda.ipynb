```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
from pathlib import Path

PROJECT_ROOT= Path.cwd().parent
OUTPUT_DIR= PROJECT_ROOT / "outputs"
```

```python
CLEAN_DATA_PATH= OUTPUT_DIR / "APL_Logistics_cleaned.csv"

df= pd.read_csv(
    CLEAN_DATA_PATH,
    encoding="latin1"
)
```

```python
print("Rows:", df.shape[0])
print("Colimns:", df.shape[1])
```

```python
missing_count= df.isnull().sum()
missing_count[missing_count > 0]
```

```python
df["Delay_Gap"]= (
    df["Days for shipping (real)"]
    - df["Days for shipment (scheduled)"]
)

df["Delay_Gap"]
```

```python
df["Delay_Gap"].describe()
```

```python
df["Delay_Gap"].median()
```

```python
def classify_delay(delay):
    if delay > 0:
        return "Delayed"
    elif delay < 0:
        return "Early"
    else:
        return "On-time"
    
df["Delay_Category"]= df["Delay_Gap"].apply(
    classify_delay
)
```

```python
df["Delay_Category"].value_counts()
```

```python
delay_distribution= (
    df["Delay_Category"]
    .value_counts(normalize=True)
    .mul(100)
    .round(2)
)
delay_distribution
```

```python
plt.figure(figsize=(13,7))

sns.countplot(
    data=df,
    x="Delay_Category",
    order=["Early", "On-time","Delayed"]
)

plt.title("Shipment Delay Classification")
plt.xlabel("Delay Category")
plt.ylabel("Number of Shipments")
plt.show()
```

```python
plt.figure(figsize=(13,7))

sns.histplot(
    df["Delay_Gap"],
    bins=20,
    kde=True
)

plt.axvline(
    0,
    linestyle= "--",
    label= "On Schedule"
)

plt.title("Distribution od Delay Gap")
plt.xlabel("Delay_Gap")
plt.ylabel("Number of shipments")
plt.legend()
plt.show()
```

```python
df["Delivery Status"].value_counts()
```

```python
delivery_staus_pct= (
    df["Delivery Status"]
    .value_counts(normalize=True)
    .mul(100)
    .round(2)
)

delivery_staus_pct
```

```python
df["SLA_Compliant"]= (
    df["Days for shipping (real)"]
    <= df["Days for shipment (scheduled)"]
)
```

```python
sla_complaince_rate= (
    df["SLA_Compliant"].mean() * 100
)

sla_complaince_rate
```

```python
late_delivery_rate= (
    (df["Delivery Status"] == "Late delivery").mean() * 100
)

late_delivery_rate
```

```python
shipping_mode_summary= (
    df.groupby("Shipping Mode")
    .agg(
        shipments=("Shipping Mode", "size"),
        Average_Actual_Days=(
            "Days for shipping (real)", "mean"
        ),
        Average_Scheduled_Days=(
            "Days for shipment (scheduled)", "mean"
        ),
        Average_Delay_Gap= (
            "Delay_Gap", "mean"
        ),
        Late_Risk_Rate= (
            "Late_delivery_risk", "mean"
        )
    ).reset_index()
)
```

```python
shipping_mode_summary["Late_Risk_Rate"]= (
    shipping_mode_summary["Late_Risk_Rate"] * 100
).round(2)

shipping_mode_summary
```

```python
shipping_mode_summary.sort_values(
    "Late_Risk_Rate",
    ascending=False
)
```

```python
plt.figure(figsize=(13,7))

sns.barplot(
    data= shipping_mode_summary.sort_values(
        "Late_Risk_Rate",
        ascending=False
    ),
    x="Shipping Mode",
    y="Late_Risk_Rate"
)

plt.title("Late Delivary Risk by Shipping Mode")
plt.xlabel("Shipping Mode")
plt.ylabel("Late Risk (%)")
plt.show()
```

```python
market_summary= (
    df.groupby("Market")
    .agg(
        Shipment= ("Market", "size"),
        Average_Delay=("Delay_Gap","mean"),
        Delayed_Rate=(
            "Delay_Category",
            lambda x: (x == "Delayed").mean() * 100
        ),
        Late_Risk_Rate= (
            "Late_delivery_risk",
            "mean"
        )
    ).reset_index()
)
```

```python
market_summary["Late_Risk_Rate"]= (
    market_summary["Late_Risk_Rate"] * 100
).round(2)

market_summary
```

```python
segment_summary= (
    df.groupby("Customer Segment")
    .agg(
        Shipment= ("Customer Segment", "size"),
        Average_Delay=("Delay_Gap","mean"),
        Delayed_Rate=(
            "Delay_Category",
            lambda x: (x == "Delayed").mean() * 100
        ),
        Late_Risk_Rate= (
            "Late_delivery_risk",
            "mean"
        )
    ).reset_index()
)

segment_summary["Late_Risk_Rate"] *= 100
segment_summary
```

```python
region_summary= (
    df.groupby("Order Region")
    .agg(
        Shipments= ("Order Region","size"),
        Average_Delay= ("Delay_Gap","mean"),
        Late_Risk_Rate= (
            "Late_delivery_risk",
            "mean"
        )
    ).reset_index()
)

region_summary["Late_Risk_Rate"] *= 100

region_summary= region_summary.sort_values(
    "Late_Risk_Rate",
    ascending=False
)

region_summary
```

```python
region_summary.head(10)
```

```python
plt.figure(figsize=(13,7))

sns.scatterplot(
    data=region_summary,
    x="Shipments",
    y="Late_Risk_Rate",
    s=100
)

plt.title("Regional Delivery Risk VS Shipment Volume")
plt.xlabel("Number of Shipment")
plt.ylabel("Late Risk (%)")
plt.show()
```

```python
country_summary= (
    df.groupby("Order Country")
    .agg(
        shipments= ("Order Country","size"),
        Late_Risk_Rate= (
            "Late_delivery_risk",
            "mean"
        ),
        Average_Delay= (
        "Delay_Gap","mean"
        )
    ).reset_index()
)
```

```python
country_summary_500= country_summary[
    country_summary["shipments"] >= 500
].copy()
```

```python
country_summary_500["Late_Risk_Rate"] *= 100
```

```python
country_summary_500.sort_values(
    "Late_Risk_Rate",
    ascending=False
).head(10)
```

```python
profit_summary= (
    df.groupby("Late_delivery_risk")
    .agg(
        Orders= ("Late_delivery_risk","size"),
        Average_Sales= ("Sales","mean"),
        Average_Profit= ("Order Profit Per Order","mean"),
        Average_Discount= ("Order Item Discount Rate","mean")
    ).reset_index()
)

profit_summary
```

```python
negative_profit_count= (
    df["Order Profit Per Order"] < 0
).sum()

negative_profit_percentage= (
    negative_profit_count / len(df) * 100
)

print("Negative-profit orders:", negative_profit_count)
print("Percentage:", round(negative_profit_percentage,2))
```

```python
department_summary= (
    df.groupby("Department Name")
    .agg(
        Shipments= ("Department Name","size"),
        Late_Risk_Rate= ("Late_delivery_risk","mean"),
        Average_Delay= ("Delay_Gap", "mean"),
        Average_Profit= ("Order Profit Per Order","mean")
    ).reset_index()
)

department_summary["Late_Risk_Rate"] *= 100

department_summary.sort_values(
    "Late_Risk_Rate",
    ascending=False
)
```
